```python
import pandas as pd
import numpy as np
import time
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.cluster import KMeans
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Choix du modèle d'Embbedings

```python
embedding = 'all-mpnet-base-v2'

# modèles à tester :
# 'all-MiniLM-L6-v2'    Le plus léger
# 'all-MiniLM-L12-v2'   mid-performance
# 'all-mpnet-base-v2'   Le meilleur = le plus lourd
```

```python
embed_minilm = SentenceTransformer(embedding)
```

#### File load

```python
sentences = ["Mon chat est gris",
             "Le pêcheur a attrapé un poisson chat",
             "Le chat du voisin a mangé une souris",
             "Le poisson chat a mangé mon chat",
             "Les lions sont juste des gros chats",
             "Qui a mangé le pêcheur ?",
             "Qu'a mangé le chat ?",
             "L'avion est blanc"]
```

```python
start_time = time.time()
```

#### Vectorisation

```python
clean_embed = embed_minilm.encode(sentences,
                                  batch_size=64,
                                  show_progress_bar=True,
                                  convert_to_tensor=True)
print(f"Nombre d'éléments encodés avec options avancées : {len(clean_embed)}")
```

```python
result = []

for i in range(len(sentences)):
  result.append([])

for i in range(len(sentences)) :
  phrase_1 = embed_minilm.encode([sentences[i]])

  for j in range(len(sentences)) :
    phrase_2 = embed_minilm.encode([sentences[j]])
    similarities = cosine_similarity(phrase_1, phrase_2)
    result[i].append(round(similarities[0][0].item(),4))
```

```python
for ele in result:
  print(ele)
```

```python
cosine_sim_matrix = np.array(result)
```

```python
# Créer une heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(cosine_sim_matrix, annot=True, cmap='coolwarm', fmt=".4f", linewidths=.5,
            xticklabels=sentences, yticklabels=sentences)
plt.title(f'Heatmap de la similarité cosinus par {embedding}')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```
